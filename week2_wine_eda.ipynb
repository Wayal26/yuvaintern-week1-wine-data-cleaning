# Week 2: Exploratory Data Analysis and Visualization

Wine Recognition dataset analysis using Pandas and Matplotlib.

```
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

wine = load_wine(as_frame=True)
df = wine.frame.copy()
df['target_name'] = df['target'].map(dict(enumerate(wine.target_names)))
df.head()
```

## 1. Dataset overview

```
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

## 2. Descriptive statistics

```
df.describe().T
```

## 3. Class distribution

```
df['target_name'].value_counts().sort_index().plot(kind='bar', title='Wine Class Distribution')
plt.xlabel('Wine Class'); plt.ylabel('Number of Samples'); plt.xticks(rotation=0); plt.show()
```

## 4. Feature distributions

```
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['alcohol'], bins=15)
axes[0].set_title('Alcohol Distribution'); axes[0].set_xlabel('Alcohol'); axes[0].set_ylabel('Frequency')
axes[1].hist(df['malic_acid'], bins=15)
axes[1].set_title('Malic Acid Distribution'); axes[1].set_xlabel('Malic Acid'); axes[1].set_ylabel('Frequency')
plt.tight_layout(); plt.show()
```

## 5. Features by class

```
df.boxplot(column='alcohol', by='target_name')
plt.suptitle(''); plt.title('Alcohol by Wine Class'); plt.xlabel('Wine Class'); plt.ylabel('Alcohol'); plt.show()

df.boxplot(column='flavanoids', by='target_name')
plt.suptitle(''); plt.title('Flavanoids by Wine Class'); plt.xlabel('Wine Class'); plt.ylabel('Flavanoids'); plt.show()
```

## 6. Correlation analysis

```
numeric = df.drop(columns=['target','target_name'])
corr = numeric.corr()
plt.figure(figsize=(10,8)); plt.imshow(corr, aspect='auto'); plt.colorbar(label='Correlation')
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90, fontsize=7)
plt.yticks(range(len(corr.columns)), corr.columns, fontsize=7)
plt.title('Correlation Heatmap'); plt.tight_layout(); plt.show()
```

## 7. Relationship between important features

```
for cls in sorted(df['target'].unique()):
    part = df[df['target'] == cls]
    plt.scatter(part['alcohol'], part['flavanoids'], label=wine.target_names[cls], alpha=0.75)
plt.title('Alcohol vs Flavanoids by Wine Class'); plt.xlabel('Alcohol'); plt.ylabel('Flavanoids'); plt.legend(); plt.show()
```

## 8. Key observations

- The dataset has 178 observations and 13 numerical predictors plus the target.
- There are no missing values or duplicate rows.
- The three wine classes are not perfectly balanced.
- Alcohol and flavanoids show useful differences across classes.
- Several feature pairs have strong positive or negative correlations, indicating redundancy and useful structure for later modeling.
